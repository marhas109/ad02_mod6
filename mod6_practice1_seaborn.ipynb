```python
# Import packages
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Import data

penguins = sns.load_dataset("penguins")
```

```python
penguins.info()
```

1. Recreate the scatter plot image you see below using Seaborn.  Use ‘flipper_length_mm’ on the x-axis and ‘body_mass_g’ on the y-axis.  The dataset called ‘penguins’ has already been imported for you in the first cell.  


scatterPlot.png

```python
sns.set(style='white')
ax = sns.scatterplot(x=penguins['flipper_length_mm'], y=penguins['body_mass_g'])

ax.set(xlabel='flipper_length_mm', ylabel='body_mass_g')

plt.show()
```

2. Using the same code from question 1, add an additional ‘hue’ argument and set that equal to the variable ‘species’.  You should get the image you see below displaying three distinct colors.

```python
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='species')

plt.show()
```

3. Now add one more argument, ‘size’, and set it equal to the variable ‘bill_length_mm’.  Also, add an additional argument, ‘style’, and set that equal to the variable ‘species’.   You should get the visual below

```python
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='species', size='bill_length_mm', style='species')

plt.show()
```
